```python

import pandas as pd

# Create a dataset
data = {
    "Name": ["Anu", "Ravi", "Priya", "Rahul", "Sneha"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 78, 88, 95]
}

df = pd.DataFrame(data)

# Display the complete dataset
print("Complete Dataset:")
print(df)

# Select the Name column using loc
print("\nName Column using loc:")
print(df.loc[:, "Name"])

# Select a specific row using loc
print("\nThird Student using loc:")
print(df.loc[2])

# Select the first column using iloc
print("\nFirst Column using iloc:")
print(df.iloc[:, 0])

# Select the third row using iloc
print("\nThird Student using iloc:")
print(df.iloc[2])

# Select Name and Marks columns using loc
print("\nName and Marks Columns:")
print(df.loc[:, ["Name", "Marks"]])

```
