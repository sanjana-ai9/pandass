```python
# DAY 41: CUSTOMER SEGMENTATION AND PRODUCT RECOMMENDATION

# STEP 1: IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics.pairwise import cosine_similarity


print("Libraries imported successfully!")


# STEP 2: CREATE CUSTOMER TRANSACTION DATASET

data = {
    'CustomerID': [101, 101, 102, 102, 103, 103, 104, 104, 105, 105,
                   106, 106, 107, 107, 108, 108, 109, 109, 110, 110],

    'Product': ['Laptop', 'Mouse', 'Mobile', 'Headphones', 'Laptop',
                'Keyboard', 'Mobile', 'Mouse', 'Tablet', 'Headphones',
                'Laptop', 'Keyboard', 'Mobile', 'Tablet', 'Laptop',
                'Mouse', 'Tablet', 'Keyboard', 'Mobile', 'Headphones'],

    'Quantity': [1, 2, 1, 1, 1, 2, 1, 3, 1, 2,
                 1, 1, 2, 1, 1, 2, 1, 2, 1, 1],

    'Price': [50000, 500, 25000, 2000, 50000,
              1500, 25000, 500, 30000, 2000,
              50000, 1500, 25000, 30000, 50000,
              500, 30000, 1500, 25000, 2000],

    'Date': pd.date_range(start='2026-01-01', periods=20, freq='D')
}

df = pd.DataFrame(data)

print("\nDataset created successfully!")

print(df.head())


# STEP 3: EXPLORE DATASET

print("\nDataset Information:")

print(df.info())

print("\nDataset Shape:")

print(df.shape)

print("\nMissing Values:")

print(df.isnull().sum())


# STEP 4: DATA CLEANING

# Remove duplicate records

df = df.drop_duplicates()

# Convert Date column into datetime format

df['Date'] = pd.to_datetime(df['Date'])

# Calculate total purchase amount

df['TotalAmount'] = df['Quantity'] * df['Price']


# STEP 5: DISPLAY CLEANED DATASET

print("\nCleaned Dataset:")

print(df.head(10))

print("\nTotal Customers:")

print(df['CustomerID'].nunique())

print("\nTotal Transactions:")

print(len(df))

print("\nTotal Sales:")

print(df['TotalAmount'].sum())

print("\nFinal Dataset Shape:")

print(df.shape)

print("\nDAY 41 - STEPS 1 TO 4 COMPLETED SUCCESSFULLY!")
```
